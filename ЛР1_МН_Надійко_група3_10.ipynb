### Завдання 25


Додатково знайдіть сам кут у градусах.

````python
a = np.array([1, 2])
b = np.array([2, 1])

cos_angle = np.dot(a, b) / (np.linalg.norm(a) * np.linalg.norm(b))

angle = np.degrees(np.arccos(cos_angle))

print("Косинус кута:", cos_angle)
print("Кут у градусах:", angle)
````

### Завдання 33
Нормалізуйте **кожен стовпець** матриці:

```text
A = [[10, 20, 30],
     [20, 30, 40],
     [30, 40, 50]]
```

за формулою:

\[
z = \frac{x - \mu}{\sigma}
\]

де `μ` — середнє значення стовпця, `σ` — стандартне відхилення стовпця.

Не використовуйте цикл `for`.

````python
A = np.array([
    [10, 20, 30],
    [20, 30, 40],
    [30, 40, 50]
])

mean = np.mean(A, axis=0)
std = np.std(A, axis=0)

Z = (A - mean) / std

print("Середні значення стовпців:", mean)
print("Стандартні відхилення:", std)

print("\nНормалізована матриця:")
print(Z)
````

### Завдання 36
Дано результати продажів трьох товарів за чотири місяці:

```text
sales = [[120, 150, 170, 160],
         [80,  90,  100, 110],
         [200, 180, 210, 230]]
```

Рядки — товари, стовпці — місяці.

Визначте:
1. загальний обсяг продажів кожного товару;
2. загальний обсяг продажів у кожному місяці;
3. товар із найбільшими сумарними продажами;
4. місяць із найбільшим загальним обсягом продажів;
5. середній обсяг продажів кожного товару.

````python
sales = np.array([
    [120, 150, 170, 160],
    [80, 90, 100, 110],
    [200, 180, 210, 230]
])

# 1. Загальний обсяг продажів кожного товару
product_sales = np.sum(sales, axis=1)

# 2. Загальний обсяг продажів у кожному місяці
monthly_sales = np.sum(sales, axis=0)

# 3. Товар із найбільшими продажами
best_product = np.argmax(product_sales) + 1

# 4. Місяць із найбільшими продажами
best_month = np.argmax(monthly_sales) + 1

# 5. Середній обсяг продажів кожного товару
average_sales = np.mean(sales, axis=1)

print("Продажі кожного товару:", product_sales)
print("Продажі кожного місяця:", monthly_sales)
print("Товар із найбільшими продажами: №", best_product)
print("Місяць із найбільшими продажами: №", best_month)
print("Середні продажі кожного товару:", average_sales)
````

### Завдання 37
Створіть квадратну матрицю `5 × 5` із випадкових цілих чисел від `1` до `20`.

Виконайте:
1. знайдіть суму головної діагоналі;
2. знайдіть суму побічної діагоналі;
3. знайдіть максимальний елемент кожного рядка;
4. знайдіть мінімальний елемент кожного стовпця;
5. замініть усі елементи, менші за середнє значення всієї матриці, на `0`.

````python
np.random.seed(42)

A = np.random.randint(1, 21, (5, 5))

print("Початкова матриця:")
print(A)

# 1. Сума головної діагоналі
main_diagonal_sum = np.trace(A)

# 2. Сума побічної діагоналі
secondary_diagonal_sum = np.trace(np.fliplr(A))

# 3. Максимальний елемент кожного рядка
row_max = np.max(A, axis=1)

# 4. Мінімальний елемент кожного стовпця
column_min = np.min(A, axis=0)

# 5. Заміна елементів, менших за середнє, на 0
mean_value = np.mean(A)
result = A.copy()
result[result < mean_value] = 0

print("\nСума головної діагоналі:", main_diagonal_sum)
print("Сума побічної діагоналі:", secondary_diagonal_sum)
print("Максимум кожного рядка:", row_max)
print("Мінімум кожного стовпця:", column_min)
print("Середнє значення:", mean_value)

print("\nМатриця після заміни:")
print(result)
````

### Завдання 39
Створіть матрицю `4 × 4` із довільних цілих чисел.

Для кожного рядка знайдіть його норму та створіть нову матрицю, у якій кожен рядок нормалізований до одиничної довжини.

### Завдання 40
Створіть матрицю коефіцієнтів і вектор правих частин для системи:

```text
 x + 2y +  z =  9
2x +  y -  z =  8
3x -  y + 2z =  3
```

1. Розв’яжіть систему за допомогою `np.linalg.solve()`.
2. Перевірте знайдений розв’язок матричним множенням `A @ x`.
3. Порівняйте результат із вектором `b`.

````python
A = np.array([
    [1, 2, 1],
    [2, 1, -1],
    [3, -1, 2]
])

b = np.array([9, 8, 3])

# 1. Розв'язання системи
x = np.linalg.solve(A, b)

print("Розв'язок системи:")
print(x)

# 2. Перевірка A @ x
result = A @ x

print("\nA @ x:")
print(result)

# 3. Порівняння з b
print("\nВектор b:")
print(b)

print("\nЧи дорівнює A @ x вектору b:")
print(np.allclose(result, b))
````
